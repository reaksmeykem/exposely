```python
# Load model directly
from transformers import AutoTokenizer, AutoModelForTextToWaveform

tokenizer = AutoTokenizer.from_pretrained("facebook/mms-tts-khm")
model = AutoModelForTextToWaveform.from_pretrained("facebook/mms-tts-khm")
```

```python
import torch
from IPython.display import Audio

# Ensure we use GPU if available
device = "cuda" if torch.cuda.is_available() else "cpu"
model.to(device)

text = "សូមជម្រាបសួរ តើអ្នកសុខសប្បាយជាទេ?" # Hello, how are you in Khmer
inputs = tokenizer(text, return_tensors="pt").to(device)

with torch.no_grad():
    output = model(**inputs).waveform

# Display the audio
Audio(output.cpu().numpy(), rate=model.config.sampling_rate)
```

```python
import torch
from openvoice import se_extractor
from openvoice.api import ToneColorConverter

# 1. Setup the Converter
device = "cuda" if torch.cuda.is_available() else "cpu"
tone_color_converter = ToneColorConverter(f'{ckpt_base}/converter/config.json', device=device)
tone_color_converter.load_ckpt(f'{ckpt_base}/converter/checkpoint.pth')

# 2. Extract the 'Voice Color' from your uploaded file
reference_speaker = '/content/Voxcpm2.m4a'
target_se, audio_name = se_extractor.get_se(reference_speaker, tone_color_converter, target_dir='processed', vad=True)

print("Voice features extracted from Voxcpm2 successfully.")
```

Now we will take the native Khmer audio we generated earlier and 'convert' its tone to match your uploaded voice.

```python
output_cloned_path = 'final_khmer_cloned_vox.wav'
# We use the previous native Khmer output as the base
source_audio = 'khmer_mms_fixed.wav'

# Get the source 'voice color' (the native Khmer base)
# Note: This is an experimental cross-lingual mapping
source_se, _ = se_extractor.get_se(source_audio, tone_color_converter, target_dir='processed', vad=True)

# Convert the tone of the Khmer speech to your voice
tone_color_converter.convert(
    batch=False,
    src_path=source_audio,
    src_se=source_se,
    tgt_se=target_se,
    output_path=output_cloned_path
)

print("Conversion complete!")
from IPython.display import Audio
display(Audio(output_cloned_path))
```

```python
import torch
from IPython.display import Audio
from transformers import AutoTokenizer, AutoModelForTextToWaveform
import os

# Configuration
model_id = "facebook/mms-tts-khm"
device = "cuda" if torch.cuda.is_available() else "cpu"
text_khmer = "សូមជម្រាបសួរ តើអ្នកសុខសប្បាយជាទេ?"
output_path = "khmer_mms_fixed.wav"

try:
    print(f"Loading native Khmer model: {model_id} on {device}...")
    tokenizer = AutoTokenizer.from_pretrained(model_id)
    model = AutoModelForTextToWaveform.from_pretrained(model_id).to(device)

    print("Synthesizing Khmer text...")
    inputs = tokenizer(text_khmer, return_tensors="pt").to(device)

    with torch.no_grad():
        output = model(**inputs).waveform

    # Move to CPU and convert to numpy for display
    audio_data = output.cpu().numpy().squeeze()
    sampling_rate = model.config.sampling_rate

    print("--- Native Khmer Synthesis Complete ---")
    display(Audio(audio_data, rate=sampling_rate))

    # Optionally save to file
    import scipy.io.wavfile as wavfile
    wavfile.write(output_path, sampling_rate, audio_data)

except Exception as e:
    print(f"Synthesis Error: {e}")
    import traceback
    traceback.print_exc()
```
